```python
import pandas as pd
import os
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

print("Libraries loaded!")
```

```python
os.chdir('..')
print(f"Working directory: {os.getcwd()}")
```

```python
# Load cleaned data from Phase 2
df_train = pd.read_csv('data/processed/train_cleaned.csv')
df_val = pd.read_csv('data/processed/val_cleaned.csv')
df_test = pd.read_csv('data/processed/test_cleaned.csv')

print(f"Train: {len(df_train)} rows")
print(f"Validation: {len(df_val)} rows")
print(f"Test: {len(df_test)} rows")
```

```python
# Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer(
    max_features=10000,  # Keep top 10,000 words only
    ngram_range=(1, 2),  # Use single words AND pairs
    min_df=2,            # Ignore words appearing in less than 2 documents
    max_df=0.95          # Ignore words appearing in more than 95% of documents
)

print("TF-IDF Vectorizer initialized!")
```

```python
# Fit TF-IDF on train data and transform all splits
X_train = tfidf.fit_transform(df_train['cleaned_text'].fillna(''))
X_val = tfidf.transform(df_val['cleaned_text'].fillna(''))
X_test = tfidf.transform(df_test['cleaned_text'].fillna(''))

print(f"Train matrix shape: {X_train.shape}")
print(f"Validation matrix shape: {X_val.shape}")
print(f"Test matrix shape: {X_test.shape}")
```

```python
# Extract mood labels
y_train = df_train['mood']
y_val = df_val['mood']
y_test = df_test['mood']

print("Label distribution in train:")
print(y_train.value_counts())
```

```python
import scipy.sparse as sp

# Save TF-IDF matrices
sp.save_npz('data/processed/X_train.npz', X_train)
sp.save_npz('data/processed/X_val.npz', X_val)
sp.save_npz('data/processed/X_test.npz', X_test)

# Save labels
y_train.to_csv('data/processed/y_train.csv', index=False)
y_val.to_csv('data/processed/y_val.csv', index=False)
y_test.to_csv('data/processed/y_test.csv', index=False)

print("All matrices and labels saved!")
```
